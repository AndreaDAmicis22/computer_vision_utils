```python
from stateful_auto_anomaly.static_stad import Config, DynamicParameters
from stateful_auto_anomaly.feature_extractor import FeatureExtractor, Providers
from stateful_auto_anomaly.metrics_computer import MahalanobisMetric
from auto_anomaly_example import AutoAnomalyExample
from pathlib import Path
import cv2 as cv
```

```python
config = Config(
    warmup_window_size=5,
    target_image_size=512,
    feature_extractor=FeatureExtractor(
        "model_path",
        providers=Providers.OPENVINO_GPU,
    ),
    metrics_computer=MahalanobisMetric(),
)
dynamic_parameters=DynamicParameters(
        area_threshold=5000, disc_threshold=30, anomaly_threshold=0.5
    )
```

```python
aa_example = AutoAnomalyExample(configuration=config, dynamic_parameters=dynamic_parameters)
aa_example.try_load_state(Path("state_path"))
```

```python
res = aa_example.predict(cv.imread("image_path"))
```
